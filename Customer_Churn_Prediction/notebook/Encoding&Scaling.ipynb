```python
# So, we have to scale the columns "tenure", "MonthlyCharges" and "TotalCharges" because they are not in the same range.
```

```python
# Another columns all value are in the range of 0 to 1 so, we apply Min-Max Scaling on this above columns.
```

```python
from sklearn.preprocessing import MinMaxScaler
MMS = MinMaxScaler()
MMS.fit(df[["tenure","MonthlyCharges","TotalCharges"]])
df[["tenure","MonthlyCharges","TotalCharges"]] = MMS.transform(df[["tenure","MonthlyCharges","TotalCharges"]])      
```
